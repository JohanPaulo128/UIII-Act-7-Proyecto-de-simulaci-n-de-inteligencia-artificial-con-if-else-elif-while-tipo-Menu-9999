```python
print("🏥 BIENVENIDO AL ASISTENTE DE ENFERMERÍA IA 🏥")
estado = "inicio"

while estado != "salir":
    print("\n--- MENÚ PRINCIPAL ---")
    print("1. Reportar malestar")
    print("2. Recomendación de higiene")
    print("3. Salir")

    opcion = input("Selecciona una opción (1-3): ")

    if opcion == "1":
        sintoma = input("¿Qué sientes? (fiebre / dolor / cansancio): ")
        if sintoma == "fiebre":
            print("IA: Te recomiendo descansar y beber muchos líquidos.")
        elif sintoma == "dolor":
            print("IA: Si es intenso, por favor acude a urgencias.")
        elif sintoma == "cansancio":
            print("IA: Es importante dormir al menos 8 horas diarias.")
        else:
            print("IA: No reconozco ese síntoma, consulta a un médico.")
    elif opcion == "2":
        print("IA: Lávate las manos frecuentemente con agua y jabón.")
    elif opcion == "3":
        estado = "salir"
        print("¡Cuídate mucho! Cerrando sistema...")
    else:
        print("Opción no válida, intenta de nuevo.")
```
